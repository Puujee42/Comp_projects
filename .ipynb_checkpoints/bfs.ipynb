```python
from pyamaze import maze,agent,COLOR,textLabel
import time 
def BFS(m):
    start=(m.rows,m.cols)
    explored=[start]
    frontier=[start]
    bfsPath={}
    goal = (1,1)
    reach = False
    while len(frontier)>0:
        currCell=frontier.pop(0)
        if currCell==goal:
            reach = True
            break
        for d in 'ESNW':
            if m.maze_map[currCell][d]==True:
                if d=='E':
                    childCell=(currCell[0],currCell[1]+1)
                if d=="S":
                    childCell=(currCell[0]+1,currCell[1])
                if d=="N":
                    childCell=(currCell[0]-1,currCell[1])
                if d=="W":
                    childCell=(currCell[0],currCell[1]-1)   
                if childCell in explored:
                    continue
                explored.append(childCell)
                frontier.append(childCell)
                bfsPath[childCell]=currCell
    fwdPath={}
    if reach:
        cell = goal 
        while cell!=start:
            fwdPath[bfsPath[cell]]=cell
            cell=bfsPath[cell]
    path_len = len(fwdPath) + 1 if reach else 0
    visited_count = len(explored)
    return reach,path_len,visited_count,fwdPath
if __name__=="__main__":
    m=maze(25,25)
    m.CreateMaze(loopPercent=100)
    start_time = time.perf_counter()
    reach , path_len,visited_count,path = BFS(m)
    end_time = time.perf_counter()
    exect_ms = (end_time - start_time) *1000 
    print("-----------------------------------------------------")
    print(f"Algorithm | Found | Path | Visited | Time (ms)")
    print(f"BFS       | {reach}  | {path_len:<4} | {visited_count:<7} | {exect_ms:.4f} ms")
    print("-----------------------------------------------------")
    if reach:
        a = agent(m, footprints=True)
        m.tracePath({a: path})
        textLabel(m, 'Path Length', path_len)
        textLabel(m, 'Visited Cells', visited_count)
        textLabel(m, 'Time (ms)', round(exect_ms, 3))
        m.run()
```
