```python
from pyamaze import maze,agent,COLOR,textLabel
import time
def DFS(m):
    start=(m.rows,m.cols)
    explored=[start]
    frontier=[start]
    dfsPath={}
    found = False
    goal = (1,1)
    while len(frontier)>0:
        currCell=frontier.pop()
        if currCell==goal:
            found =  True
            break
        for d in 'ESNW':
            if m.maze_map[currCell][d]==True:
                if d=='E':
                    childCell=(currCell[0],currCell[1]+1)
                if d=="S":
                    childCell=(currCell[0]+1,currCell[1])
                if d=="N":
                    childCell=(currCell[0]-1,currCell[1])
                if d=="W":
                    childCell=(currCell[0],currCell[1]-1)   
                if childCell in explored:
                    continue
                explored.append(childCell)
                frontier.append(childCell)
                dfsPath[childCell]=currCell
    fwdPath={}
    if found:
        cell=goal
        while cell!=start:
            fwdPath[dfsPath[cell]]=cell
            cell=dfsPath[cell]
    path_len = len(fwdPath)
    visited_len = len(explored)
    return found,path_len,visited_len,fwdPath
if __name__=="__main__":
    m=maze(25,25)
    m.CreateMaze(loopPercent=100)
    start_time = time.perf_counter()
    found,path_len,visited_len,path=DFS(m)
    end_time = time.perf_counter()
    exect_ms = (end_time-start_time)*1000
    print("-----------------------------------------------------")
    print(f"Algorithm | Found | Path | Visited | Time (ms)")
    print(f"DFS       | {found}  | {path_len:<4} | {visited_len:<7} | {exect_ms:.4f} ms")
    print("-----------------------------------------------------")
    if found:
        a = agent(m, footprints=True)
        m.tracePath({a: path})
        textLabel(m, 'Path Length', path_len)
        textLabel(m, 'Visited Cells', visited_len)
```
