```python
from pyamaze import maze, agent, COLOR, textLabel
from queue import PriorityQueue
import time

def h(cell1, cell2):
    """Manhattan distance heuristic function."""
    return abs(cell1[0] - cell2[0]) + abs(cell1[1] - cell2[1])

def GBFS(m):
    start = (m.rows, m.cols)
    goal = (1, 1)

    frontier = PriorityQueue()
    frontier.put((h(start, goal), start))

    explored = [start]
    explored_set = {start}
    gbfsPath = {}
    reach = False

    while not frontier.empty():
        _, currCell = frontier.get()

        if currCell == goal:
            reach = True
            break

        for d in 'ESNW':
            if m.maze_map[currCell][d] == True:
                if d == 'E':
                    childCell = (currCell[0], currCell[1] + 1)
                elif d == 'S':
                    childCell = (currCell[0] + 1, currCell[1])
                elif d == 'N':
                    childCell = (currCell[0] - 1, currCell[1])
                elif d == 'W':
                    childCell = (currCell[0], currCell[1] - 1)

                if childCell in explored_set:
                    continue

                explored.append(childCell)
                explored_set.add(childCell)
                
                frontier.put((h(childCell, goal), childCell))
                gbfsPath[childCell] = currCell

    fwdPath = {}
    if reach:
        cell = goal
        while cell != start:
            fwdPath[gbfsPath[cell]] = cell
            cell = gbfsPath[cell]

    path_len = len(fwdPath) + 1 if reach else 0
    visited_count = len(explored)
    return reach, path_len, visited_count, fwdPath

if __name__ == "__main__":
    m = maze(25, 25)
    m.CreateMaze(loopPercent=100)

    start_time = time.perf_counter()
    reach, path_len, visited_count, path = GBFS(m)
    end_time = time.perf_counter()
    exect_ms = (end_time - start_time) * 1000

    print("-----------------------------------------------------")
    print(f"Algorithm | Found | Path | Visited | Time (ms)")
    print(f"GBFS      | {reach}  | {path_len:<4} | {visited_count:<7} | {exect_ms:.4f} ms")
    print("-----------------------------------------------------")

    if reach:
        a = agent(m, footprints=True, color=COLOR.cyan)
        m.tracePath({a: path})
        textLabel(m, 'Algorithm', 'GBFS')
        textLabel(m, 'Path Length', path_len)
        textLabel(m, 'Visited Cells', visited_count)
        textLabel(m, 'Time (ms)', round(exect_ms, 3))
        m.run()
```
