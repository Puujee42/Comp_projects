```python
from pyamaze import maze, agent, COLOR, textLabel
from queue import PriorityQueue
import time
def h(cell1, cell2):
    x1, y1 = cell1
    x2, y2 = cell2
    return abs(x1 - x2) + abs(y1 - y2)

def aStar(m):
    start = (m.rows, m.cols)
    found = False
    goal = (1,1)
    g_score = {cell: float('inf') for cell in m.grid}
    g_score[start] = 0
    f_score = {cell: float('inf') for cell in m.grid}
    f_score[start] = h(start, (1, 1))
    
    open = PriorityQueue()
    aPath = {}
    open.put((h(start, (1, 1)), h(start, (1, 1)), start))
    visited = set()
    while not open.empty():
        currCell = open.get()[2]
        visited.add(currCell)
        if currCell == goal:
            found = True
            break
        for d in "ESNW":
            if m.maze_map[currCell][d] == True:
                if d == 'E':
                    childCell = (currCell[0], currCell[1] + 1)
                elif d == "S":
                    childCell = (currCell[0] + 1, currCell[1])
                elif d == "N":
                    childCell = (currCell[0] - 1, currCell[1])
                elif d == "W":
                    childCell = (currCell[0], currCell[1] - 1) 
                    
                temp_g_score = g_score[currCell] + 1
                temp_f_score = temp_g_score + h(childCell, (1, 1))
                
                if temp_f_score < f_score[childCell]:
                    g_score[childCell] = temp_g_score
                    f_score[childCell] = temp_f_score
                    open.put((temp_f_score, h(childCell, (1, 1)), childCell))
                    aPath[childCell] = currCell
                    
    fwdPath = {}
    if found :
        cell = goal
        while cell != start:
            fwdPath[aPath[cell]] = cell
            cell = aPath[cell]
    path_len = len(fwdPath) + 1 if found else 0
    visited_count = len(visited)
    return found,path_len,visited_count,fwdPath

if __name__ == "__main__":
    m = maze(25,25)
    m.CreateMaze(loopPercent=100)
    start_time = time.perf_counter()
    found,path_len,visited_count,path = aStar(m)
    end_time = time.perf_counter()
    ms = (end_time - start_time) * 1000
    print("-----------------------------------------------------")
    print("Algorithm | Found | Path | Visited | Time (ms)")
    print(f"A*        | {found}  | {path_len:<4} | {visited_count:<7} | {ms:.4f} ms")
    print("-----------------------------------------------------")
    if found:
        a = agent(m, footprints=True, color=COLOR.blue)
        m.tracePath({a: path})
        textLabel(m, 'A* Path Length', path_len)
        textLabel(m, 'Visited Cells', visited_count)
        textLabel(m, 'Time (ms)', round(ms, 3))
        m.run()
```
